```python
import json

INPUT = "arxiv-metadata-oai-snapshot.json"
OUTPUT = "clear_arxiv_data_mathAG.json"

CATEGORIES = set([
    "math.AG",   # ядро
    "math.AC",   # коммутативная алгебра
    "math.NT",   # арифметическая геометрия
    "math.RT",   # теория представлений
    "math.KT",   # K-теория, мотивы
    "math.AT",   # алгебраическая топология
    "math.DG",   # комплексная/дифференциальная геометрия
    "math.SG",   # симплектическая геометрия
    "math.QA",   # квантовые группы
    "math-ph",   # математическая физика
    "hep-th",    # зеркальная симметрия, струны
])

n = 0

#перерабатываем файл, оставляя только нужные части(id, title, abstract) и математические темы
with open(INPUT, "r", encoding='utf-8') as f_in, open(OUTPUT, "w", encoding='utf-8') as f_out:
    for line in f_in:

        #проверка, что данные есть
        line = line.strip()
        if not line:
            continue
        try:
            rec = json.loads(line)
        except json.JSONDecodeError:
            continue


        #проверяем, что статья про математику
        cur_cat = rec.get("categories", [""])[0].split()
        is_about_math = 0
        for cat in cur_cat:
            if cat in CATEGORIES:
                is_about_math = 1
                break
        
        if is_about_math == 0:
            continue

        

        slim = {
                'id': rec.get('id', ''),
                'title': rec.get('title', ''),
                'abstract': rec.get('abstract', ''),
        }

        if not slim['title'] and not slim['abstract']:
            continue
        f_out.write(json.dumps(slim, ensure_ascii=False) + '\n')
        n += 1

print("Готово", n)
```

```python
import json

INPUT = "arxiv-metadata-oai-snapshot.json"
OUTPUT = "only_mathAG_and_mathAC.json"

CATEGORIES = set([
    "math.AG",   # ядро
    "math.AC",   # коммутативная алгебра
])

titles = set()
n = 0

#перерабатываем файл, оставляя только нужные части(id, title, abstract) и математические темы
with open(INPUT, "r", encoding='utf-8') as f_in, open(OUTPUT, "w", encoding='utf-8') as f_out:
    for line in f_in:

        #проверка, что данные есть
        line = line.strip()
        if not line:
            continue
        try:
            rec = json.loads(line)
        except json.JSONDecodeError:
            continue


        #проверяем, что статья про математику
        cur_cat = rec.get("categories", [""])[0].split()
        
        is_about_math = 0
        for cat in cur_cat:
            if cat in CATEGORIES:
                is_about_math = 1
                break
        
        if is_about_math == 0:
            continue

        

        slim = {
                'id': rec.get('id', ''),
                'title': rec.get('title', ''),
                'abstract': rec.get('abstract', ''),
        }

        if not slim['title'] and not slim['abstract'] or slim['title'] in titles:
            continue
        titles.add(slim['title'])
        f_out.write(json.dumps(slim, ensure_ascii=False) + '\n')
        n += 1

print("Готово", n)
```
