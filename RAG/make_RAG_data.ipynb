```python
def vnutryak(text: str):
    i = 0
    n = len(text)
    
    while text[i] != "{":
        i += 1

    i += 1
    start = i
    skobki = 1
    while i < n and skobki != 0:
        if text[i] == "}":
            skobki -= 1
        if text[i] == "{":
            skobki += 1
        i += 1

    return text[start:i-1]
```

```python
import re

def make_sections(filename:str):
    sections = []
    with open(filename, "r", encoding="utf-8") as f:
        text = f.read()

    text_splited = re.split(r'(?=\\section\*?\{)', text)    

    for i in range(1, len(text_splited)):
        name = vnutryak(text_splited[i].split("\n")[0])  
        section = {
            "name" : name,
            "text": text_splited[i],
            "doc_name": filename
        }
        sections.append(section)

    return sections
```
